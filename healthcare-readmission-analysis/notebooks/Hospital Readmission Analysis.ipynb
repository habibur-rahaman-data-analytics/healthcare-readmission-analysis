#### Create analysis-ready columns

```python
# To check readmission within 30 days, create a new column 'readmitted_30_days'

df['readmitted_30_days'] = (df['readmitted'] == '<30').astype('int')
df[['readmitted', 'readmitted_30_days']].value_counts(dropna=False)
```

```python
# Calculate the percentage of readmission within 30 days

readmission_rate = (
    df['readmitted_30_days'].mean() * 100
)
print(f'Recorded <30 rate: {readmission_rate:.2f}%')
```

##### Identify patient groups with higher readmission rates

```python
# To check which age groups have higher recorded 30-day readmission rates

age_analysis = (
    df.groupby('age', observed=True)
      ['readmitted_30_days']
      .agg(['mean', 'count'])
      .reset_index()
)

age_analysis['readmission_rate_pct'] = (
    age_analysis['mean'] * 100
)

age_analysis
```

```python
gender_analysis = (
    df.groupby('gender')['readmitted_30_days']
      .agg(['count', 'mean'])
      .reset_index()
)

gender_analysis['readmission_rate_pct'] = (
    gender_analysis['mean'] * 100
)

gender_analysis
```

```python
# Calculate the recorded 30-day readmission rate for each group

inpatient_analysis = (
    df.groupby('number_inpatient', observed=False)
      .agg(
          Total_Encounters=('readmitted_30_days', 'count'),
          Readmission_Rate=('readmitted_30_days', 'mean')
      )
      .reset_index()
)

inpatient_analysis['Readmission_Rate'] *= 100

inpatient_analysis.round(2)
```

```python
# To check the diabetes medication

df['medication_changed'] = (df['change'].map({'Ch': 1, 'No': 0})).astype('Int64')

df[['change', 'medication_changed']].value_counts()
```

##### Investigate clinical and admission factors

```python
# To check whether 30-day readmission varies with the length of the hospital stay

df['length_of_stay_group'] = pd.cut(
    df['time_in_hospital'],
    bins=[0, 3, 7, 14],
    labels=['1-3 days', '4-7 days', '8-14 days']
)

df['length_of_stay_group'].value_counts()
```

```python
df['prior_inpatient_group'] = pd.cut(
    df['number_inpatient'],
    bins=[-1, 0, 1, float('inf')],
    labels=['0 visits', '1 visit', '2+ visits']
)

df['prior_inpatient_group'].value_counts()
```

```python

df['prior_emergency_group'] = pd.cut(
    df['number_emergency'],
    bins=[-1, 0, 1, float('inf')],
    labels=['0 visits', '1 visit', '2+ visits']
)

df['prior_emergency_group'].value_counts()
```

##### Analyze previous hospital visits

```python
# Analyze previous hospital visits

# Select the visit columns
visit_columns = [
    'number_inpatient',
    'number_emergency',
    'number_outpatient'
]

# Calculate total previous visits for each row
df['total_previous_visits'] = df[visit_columns].sum(axis=1, min_count=1)

# Count the total_previous_visits values
df['total_previous_visits'].value_counts()
```

```python
# Creating a separate column

df['readmission_category'] = (
    df['readmitted']
    .map({
        'NO': 'No recorded readmission',
        '<30': 'Readmitted within 30 days',
        '>30': 'Readmitted after 30 days'
    })
)
df['readmission_category'].value_counts(dropna=False)
```

```python
# Calculate the distribution as percentages

category_percentage = (df['readmission_category'].value_counts(normalize=True, dropna=False).mul(100).round(2))
category_percentage
```
